**Caso DermAlpes.** DermAlpes conecta centros de salud de municipios pequeños con dermatólogos a distancia y quiere una herramienta que delimite automáticamente la lesión en cada imagen dermatoscópica, para que el especialista pueda medir su tamaño y forma. Como no tiene imágenes etiquetadas de sus pacientes, la primera idea es segmentar **sin entrenar un modelo supervisado**: agrupar los píxeles de cada imagen con K-medias según su color y su posición, y decidir de forma automática qué grupos son lesión. En este notebook construimos ese método, lo ajustamos con 15 imágenes que tienen máscara de referencia, lo evaluamos una sola vez sobre 5 imágenes reservadas y discutimos dónde falla y por qué.

**Fuente de los datos:** ISIC 2016 ([enlace](https://challenge.isic-archive.com/landing/2016/37/)), tarea de segmentación de lesiones.
